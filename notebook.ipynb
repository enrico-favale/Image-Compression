```python
# Cubo di Rubik
rubik = ImageCompressor(input_dir="Rubik", image_name="original.jpg")

# 1️⃣ Compressione leggera: wavelet avanzata, soglia bassa, 1 livello, mantieni tutti i dettagli
rubik.compress_color_image_dwt(
    wavelet="db2",
    level=2,
    threshold=50,
    keep_ll_only=False,
    show=True,
    save_name="compressed_light.jpg",
    save_coeffs=True
)

# 2️⃣ Compressione media: wavelet semplice, soglia media, 2 livelli, mantieni tutti i dettagli
rubik.compress_color_image_dwt(
    wavelet="haar",
    level=3,
    threshold=80,
    keep_ll_only=False,
    show=True,
    save_name="compressed_medium.jpg",
    save_coeffs=True
)

# 3️⃣ Compressione pesante: wavelet avanzata, soglia alta, 2 livelli, mantieni solo LL
rubik.compress_color_image_dwt(
    wavelet="db2",
    level=3,
    threshold=100,
    keep_ll_only=True,
    show=True,
    save_name="compressed_LLonly.jpg",
    save_coeffs=True
)

# 4️⃣ Compressione brutale: wavelet semplice, soglia altissima, 3 livelli, solo LL
rubik.compress_color_image_dwt(
    wavelet="haar",
    level=4,
    threshold=150,
    keep_ll_only=True,
    show=True,
    save_name="compressed_aggressive.jpg",
    save_coeffs=True
)

```
